```python
list=[]

N = int(input())

my_list = []

for _ in range(N):
    command = input().split()

    if command[0] == "insert":
        my_list.insert(int(command[1]), int(command[2]))

    elif command[0] == "print":
        print(my_list)

    elif command[0] == "remove":
        my_list.remove(int(command[1]))

    elif command[0] == "append":
        my_list.append(int(command[1]))

    elif command[0] == "sort":
        my_list.sort()

    elif command[0] == "pop":
        my_list.pop()

    elif command[0] == "reverse":
        my_list.reverse()
```

```python


n = int(input("Number of contacts: "))

phone_book = {}

for _ in range(n):
    name, phone = input("Enter name and phone: ").split()
    phone_book[name] = phone

q = int(input("Number of searches: "))

for _ in range(q):
    name = input("Search name: ")

    if name in phone_book:
        print(f"{name}={phone_book[name]}")
    else:
        print("Not found")

```

```python

data = """5
youssef
37.21
ahmed
37.21
Tina
37.2
ali
41
hossam
39"""

lines = iter(data.splitlines())

n = int(next(lines))

students = []

for _ in range(n):
    name = next(lines)
    score = float(next(lines))
    students.append([name, score])

grades = sorted(set(score for name, score in students))

second_lowest = grades[1]

result = []

for name, score in students:
    if score == second_lowest:
        result.append(name)

result.sort()

for name in result:
    print(name)
```
